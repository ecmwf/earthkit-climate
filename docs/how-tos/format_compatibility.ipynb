# Computing Climate Indicators Across Different Data Formats

This how-to guide demonstrates how **earthkit-climate** indicators accept multiple data structures natively—including **earthkit-data** `FieldList` objects, `xarray.Dataset`, and `xarray.DataArray`—without requiring manual format conversions.

**earthkit-climate** performs automatic type inspection and format conversion behind the scenes, returning CF-compliant outputs regardless of the input data type.

---

## 1. Setup and Package Imports

Import `earthkit.data` for loading sample datasets and `earthkit.climate` for indicator calculations.

```python
import earthkit.data as ekd
import xarray as xr

import earthkit.climate as ekc
```

## 2. Using an earthkit-data `FieldList` Input

Data retrieved from GRIB or NetCDF files using **earthkit-data** can be accessed as a `FieldList` object. You can pass a `FieldList` directly to any **earthkit-climate** indicator function.

```python
# Retrieve sample NetCDF dataset as an earthkit-data FieldList
source = ekd.from_source("earthkit-climate-sample", "tasmax_ACCESS-CM2_historical_reference")
field_list = source.to_fieldlist()

# Pass FieldList directly to indicator function
hot_days_fl = ekc.indicators.tx_days_above(field_list, thresh="300 K", freq="YS")
hot_days_fl
```

## 3. Using an `xarray.Dataset` Input

Indicators also accept full `xarray.Dataset` containers containing climate variable arrays.

```python
# Convert source dataset to xarray.Dataset
ds_temp = source.to_xarray()

# Pass Dataset directly to indicator function
hot_days_ds = ekc.indicators.tx_days_above(ds_temp, thresh="300 K", freq="YS")
hot_days_ds
```

## 4. Using an `xarray.DataArray` Input

If working with an individual variable extracted from a dataset as an `xarray.DataArray`, pass it directly to the indicator function.

```python
# Extract individual DataArray from Dataset
da_temp = ds_temp["tasmax"]

# Pass DataArray directly to indicator function
hot_days_da = ekc.indicators.tx_days_above(da_temp, thresh="300 K", freq="YS")
hot_days_da
```

## Summary

- **earthkit-climate** indicator functions support `FieldList`, `Dataset`, and `DataArray` inputs interchangeably.
- Type inspection and format conversion occur automatically, producing enriched CF-compliant outputs with standard metadata across all input types.
