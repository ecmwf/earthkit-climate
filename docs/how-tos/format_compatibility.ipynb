# Climate Indicators From Different Data Formats

This how-to guide demonstrates how **earthkit-climate** indicators accept multiple data structures natively, including **earthkit-data** `FieldList` objects, `xarray.Dataset`, and `xarray.DataArray`.

**earthkit-climate** performs automatic type inspection and format conversion behind the scenes, returning CF-compliant outputs regardless of the input data type.

```python
import earthkit.data as ekd

import earthkit.climate as ekc
```

```python
source = ekd.from_source("earthkit-climate-sample", "tasmax_ACCESS-CM2_historical_reference")
```

## Xarray `DataArray`

If working with an individual variable extracted from a dataset as an `xarray.DataArray`, pass it directly to the indicator function.

```python
dataset = source.to_xarray(chunks={})
dataarray = dataset["tasmax"]

ekc.indicators.tx_days_above(dataarray, thresh="300 K", freq="YS")
```

## Xarray `Dataset`

Indicators also accept full `xarray.Dataset` containers with the `ds` argument.
The required variables for the indicator are then extracted based on default variable names or string-valued overrides as in the following example:

```python
ekc.indicators.tx_days_above(ds=dataset, tasmax="tasmax", thresh="300 K", freq="YS")
```

This form of providing data to an indicator function is particularly convenient if the indicator requires multiple inputs which are all available in a single Dataset.

## Earthkit-data `FieldList`

Data retrieved from GRIB or NetCDF files using **earthkit-data** can be accessed as a `FieldList` object. You can pass a `FieldList` directly to any **earthkit-climate** indicator function.

```python
fieldlist = source.to_fieldlist()

ekc.indicators.tx_days_above(fieldlist, thresh="300 K", freq="YS")
```

Note that an xarray DataArray is returned.

## Summary

- **earthkit-climate** indicator functions support `FieldList`, `Dataset`, and `DataArray` inputs interchangeably.
- Type inspection and format conversion occur automatically, producing enriched CF-compliant outputs with standard metadata across all input types.
